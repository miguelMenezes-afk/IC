Sigmoid neurons simulating perceptrons, part I 
Suppose we take all the weights and biases in a network of perceptrons, and multiply them by a positive constant, c>0
. Show that the behaviour of the network doesn't change.

R: Em um Perceptron, multiplicar os pesos por "c" não muda nada, mas em um neurônio sigmoid, que é uma curva suave, multiplicar os pesos por uma constante estica o eixo Z. Se "c" tender ao infinito a curva suave do sigmoide é esmagado até ficar reto igual a um Perceptron.

A forma como o Perceptron toma decisões é calculando o produto escalar das entradas($x$) pelo peso($w$) e soma do viés($b$), depois passa pela função "degrau". Chamando o resultado desse cálculo de $z$ temos:

$$z = \sum_{i} w_i x_i + b$$

Regra da ativação do Perceptron:
- Se $z \ge 0$, o Perceptron dispara 1.
- Se $z < 0$, o Perceptron distpara 0.

Criando a nova rede neural multiplicando o peso $w$ por c ($c \cdot w$)e o viés $b$ também por c ($c \cdot b$).
Chamaremos o novo valor de $z_{novo}$.

$$z_{novo} = \sum_{i} (c \cdot w_{i})x_{i} + (c \cdot b)$$

Colocando em evidência o c teremos:

$$z_{novo} = c \cdot \left( \sum_{i} w_i x_i + b \right)$$

O termo em parênteses é igual ao $z$, então iremos substituir:

$$z_{novo} = c \cdot z$$

### A resposta não muda.

Vamos aplicar a regra de decisão no $z_{novo}$ e o valor irá somente aumentar ou permanecer zero, mas o sinal nunca mudará, não vai de positivo para negativo e nem vice e versa.
- Se o original disparava 1, significa que o $z \ge 0$. Multiplicar um número maior ou igual a zero por uma constante positiva($c$), obrigatoriamente vai resultar em um número positivo; ex: $z = 2$; $c = 40$; $z_{novo} = 80$, sempre será positivo e o neurônio irá disparar 1.

- Se o original dispara 0, significa que o $z < 0$. Multiplicar um número negativo por um positivo($c$), obrigatoriamente vai resultar em um número negativo; ex: $z = -3$; $c = 4$; $z_{novo} = -12$, sempre será negativo e o neurônio irá disparar 0.
