```python
import numpy as np
import matplotlib.pyplot as plt

# definições matemáticas vetorizadas (processam arrays inteiras de uma vez)
def sigmoid(z):
    """Calcul a função sigmoide de uma forma vetorizada."""
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(z):
    """Calcula a derivada da sigmoide multiplicando a ativação por (1 - ativação)."""
    s = sigmoid(z)
    return s * (1 - s)

# 2. Simulando o espaço de entrada
# Criamos 1000 valores de z, indo de -10 a 10 (representando a extensão da nossa Curva Normal)
z_valores = np.linspace(-10, 10, 1000)

# 3. Processando os dados
ativacoes = sigmoid(z_valores)
gradientes = sigmoid_derivative(z_valores)

# 4. Plotagem Analítica
plt.figure(figsize=(10, 6))

# A Função em S
plt.plot(z_valores, ativacoes, label=r'Função Sigmoide $\sigma(z)$', color='blue', linewidth=2)

# A Derivada mostrando o esmagamento matemático
plt.plot(z_valores, gradientes, label=r"Derivada $\sigma'(z)$ (Máx = 0.25)", color='red', linestyle='--', linewidth=2)

# Linhas de referência geométrica
plt.axhline(0.25, color='gray', linestyle=':', label='Teto do Gradiente (0.25)')
plt.axvline(0, color='black', linewidth=0.8)
plt.axhline(0, color='black', linewidth=0.8)

plt.title("O Neurônio Sigmoide e o Desaparecimento do Gradiente", fontsize=14)
plt.xlabel("Input Estocástico do Neurônio (z)", fontsize=12)
plt.ylabel("Output / Valor da Derivada", fontsize=12)
plt.legend(fontsize=12)
plt.grid(alpha=0.3)
plt.show()
```
