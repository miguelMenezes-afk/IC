```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definição matemática vetorizada
def sigmoid(z):
    """Calcula a ativação Sigmoide de forma vetorizada."""
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(z):
    """Calcula a derivada da Sigmoide."""
    s = sigmoid(z)
    return s * (1 - s)

# 2. Gerando um Random Walk de dados (representando o eixo z)
# Simulamos valores de z de -10 a 10 (as caudas da nossa curva normal)
z_valores = np.linspace(-10, 10, 1000)

# 3. Calculando as funções
ativacoes = sigmoid(z_valores)
gradientes = sigmoid_derivative(z_valores)

# 4. Plotando a visualização para o seu relatório da IC
plt.figure(figsize=(10, 6))

# Curva da Função (O "S")
plt.plot(z_valores, ativacoes, label=r'Função Sigmoide $\sigma(z)$', color='blue', linewidth=2)

# Curva da Derivada (A prova do Vanishing Gradient)
plt.plot(z_valores, gradientes, label=r"Derivada $\sigma'(z)$ (Máx = 0.25)", color='red', linestyle='--', linewidth=2)

# Linhas auxiliares
plt.axhline(0.25, color='gray', linestyle=':', label='Teto do Gradiente (0.25)')
plt.axvline(0, color='black', linewidth=0.8)
plt.axhline(0, color='black', linewidth=0.8)

plt.title("O Neurônio Sigmoide e o Esmagamento do Gradiente", fontsize=14)
plt.xlabel("Input do Neurônio (z)", fontsize=12)
plt.ylabel("Output / Valor do Gradiente", fontsize=12)
plt.legend(fontsize=12)
plt.grid(alpha=0.3)
plt.show()
```
