```python
import numpy as np
import matplotlib.pyplot as plt

#1 simulando 10.000 dados com distribuição normal (média 0, variância 1)
# pense nisso como as posições finais da bola do tabuleiro de Galton
z = np.random.randn(10000)

#2 passando os dados pela função sigmoide
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

#3 plotando o antes e o depois usando histogramas
plt.figure(figsize=(12, 5))

z_valores = np.linspace(-10, 10, 1000)

ativacoes = sigmoid(z_valores)

# Eixo 1: A Curva Normal Original
plt.subplot(1, 2, 1)
plt.hist(z, bins=50, color='gray', edgecolor='black', alpha=0.7)
plt.axvline(0, color='red', linestyle='dashed', linewidth=2, label='Média = 0')
plt.title("Antes: Curva Normal Centrada")
plt.xlabel("Input (z)")
plt.ylabel("Frequência")
plt.legend()

# Eixo 2: A Destruição Estocástica pela Sigmoide
plt.subplot(1, 2, 2)
plt.hist(ativacoes, bins=50, color='blue', edgecolor='black', alpha=0.7)
plt.axvline(0.5, color='red', linestyle='dashed', linewidth=2, label='Média ≈ 0.5')
plt.title("Depois: O Efeito da Sigmoide")
plt.xlabel("Output (Ativação)")
plt.legend()

plt.tight_layout()
plt.show()
```
