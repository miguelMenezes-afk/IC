Sigmoid neurons simulating perceptrons, part II 
Suppose we have the same setup as the last problem - a network of perceptrons. Suppose also that the overall input to the network of perceptrons has been chosen. We won't need the actual input value, we just need the input to have been fixed. Suppose the weights and biases are such that w⋅x+b≠0

 for the input x
 to any particular perceptron in the network. Now replace all the perceptrons in the network by sigmoid neurons, and multiply the weights and biases by a positive constant c>0
 
. Show that in the limit as c→∞
 the behaviour of this network of sigmoid neurons is exactly the same as the network of perceptrons. How can this fail when w⋅x+b=0
 for one of the perceptrons?

Já temos o neurônio original $z = w \cdot x + b$. Substituímos $z$ pelo $z_{novo} = c \cdot z$; mas para melhorar a compreensão usaremos somente o $c \cdot z$.
Função sigmóide com a constante:

$$\sigma(c \cdot z) = \frac{1}{1 + e^{-c \cdot z}}$$

$z \neq 0$.    
$z$ é positivo ou negativo. Testaremos os dois cenários, aplicando o limite $c \to \infty$

Cenário A: z é positivo.
- Quando z é positivo no Perceptron a saída é 1. 

- No nosso sigmóide modificado o $c$ tende ao infinito positivo e $z$ é positivo, a multiplicação $c \cdot z$ resulta em $+\infty$.

- Já a parte de baixo $e^{- \infty}$ tende a 0.

- A equação se torna $\frac{1}{1 + 0} = 1$.

- Vai ser disparado 1, sendo que a vantagem do sigmóide é não ser exatamente 1 ou 0, e sim um valor entre 1 e 0, de acordo com a suavidade do gráfico.

Cenário B: z é negativo.
- Quando z é negativo no Perceptron a saída é 0.

- No nosso sigmóide modificado o $c$ tende ao infinito negativo e $z$ é negativo, a multiplicação $c \cdot z$ resulta em $-\infty$.

- Já a parte de baixo $e^{+ \infty}$ a base $e$ está sendo elevado a um expoente que já tem um sinal negativo então $(-(- \infty))$ vira $e^{+ \infty}$ que tende a $+\infty$.

- A equação se torna $\frac{1}{1 + \infty}$, que resulta em 0.

- Vai ser disparado 0, e não um valor entre 0 e 1.


Como fizemos os dois cenários possíveis com $c = + \infty$, e que os resultados serão 0 e 1 igual a um Perceptron.

Por ultimo, falaremos do $w \cdot x + b = 0$, para simplificar $z = 0$.

1. No perceptron a regra diz que $z \geq 0$ o neurônio dispara 1.

2. Na sigmoide, se $z = 0$, não importa qual o valor de $c$, pois qualquer coisa multiplicado por 0 é 0 $(c \cdot 0 = 0)$.

3. Voltemos a equação $\sigma(0) = \frac{1}{1 + e^{0}} \to \frac{1}{1 + 1} = 0,5$
