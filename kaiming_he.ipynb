```python
import numpy as np
import matplotlib.pyplot as plt

#1 dados originais(curva normal/tabuleiro de galton)
z = np.random.randn(1000)

#2 as funções de ativação
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def relu(z):
    return np.maximum(0, z)

ativacoes_sig = sigmoid(z)
ativacoes_relu = relu(z)

#3 duelo de histogramas
plt.figure(figsize=(12, 5))

# Eixo 1: A Curva Normal (Original)
plt.subplot(1, 3, 1)
plt.hist(z, bins=50, color='gray', edgecolor='black', alpha=0.7)
plt.title("1. Original: Curva Normal (Média 0)")

# Eixo 2: O Efeito Sigmoide (O seu insight)
plt.subplot(1, 3, 2)
plt.hist(ativacoes_sig, bins=50, color='blue', edgecolor='black', alpha=0.7)
plt.title("2. Sigmoide: Saturação em 0 e 1")

# Eixo 3: O Efeito ReLU (O Padrão Moderno)
plt.subplot(1, 3, 3)
plt.hist(ativacoes_relu, bins=50, color='green', edgecolor='black', alpha=0.7)
plt.title("3. ReLU: O Corte ao Meio")

plt.tight_layout()
plt.show()
```
